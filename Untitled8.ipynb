handling outliers


```python
import matplotlib.pyplot as plt
import seaborn as sns
num_cols = [
    "monthly_minutes",
    "monthly_data_mb",
    "monthly_sms",
    "contract_months",
    "overdue_payments",
]
plt.figure(figsize=(15, 8))
for i, col in enumerate(num_cols, 1):
  plt.subplot(2, 3, i)
  sns.boxplot(y=df[col])
  plt.title(f"Boxplot of {col}")
plt.tight_layout()
plt.show()
```

```python
for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    outliers = df[(df[col] < lower) | (df[col] > upper)]
    print(col, ":", len(outliers), "outliers")
```

```python
for col in num_cols:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1
    lower = Q1 - 1.5 * IQR
    upper = Q3 + 1.5 * IQR
    df[col] = df[col].clip(lower, upper)
```

```python
plt.figure(figsize=(12, 6))
sns.boxplot(data=df[num_cols])
plt.xticks(rotation=45)
plt.title("Boxplot After Outlier Handling")
plt.show()
```

Univariate analysis

HISTOGRAM,PIE chart,BOX plot


```python
import matplotlib.pyplot as plt
import seaborn as sns
num_cols = [
    "monthly_minutes",
    "monthly_data_mb",
    "monthly_sms",
    "contract_months",
    "overdue_payments"
]
for col in num_cols:
    fig, axes = plt.subplots(1, 3, figsize=(18, 5))
    sns.histplot(df[col], kde=True, ax=axes[0])
    axes[0].set_title("Histogram of " + col)
    pd.cut(df[col], bins=5).value_counts().plot(
        kind="pie", autopct="%1.1f%%", ax=axes[1]
    )
    axes[1].set_title("Pie Chart of " + col)
    axes[1].set_ylabel("")
    sns.boxplot(x=df[col], ax=axes[2])
    axes[2].set_title("Boxplot of " + col)
    plt.tight_layout()
    plt.show()
```

Bivariate analysis

COUNT plot,SCATTER plot,BOX plot



```python
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(7, 5))
sns.countplot(
    data=df,
    x="plan_type_Basic",
    hue="churn_next_month"
)
plt.title("Plan Type vs Customer Churn")
plt.xlabel("Basic Plan (0 = No, 1 = Yes)")
plt.ylabel("Customer Count")
plt.show()
plt.figure(figsize=(7, 5))
sns.scatterplot(
    data=df,
    x="monthly_minutes",
    y="churn_next_month"
)
plt.title("Monthly Minutes vs Customer Churn")
plt.xlabel("Monthly Minutes")
plt.ylabel("Churn (0 = No, 1 = Yes)")
plt.show()
plt.figure(figsize=(7, 5))
sns.boxplot(
    data=df,
    x="churn_next_month",
    y="overdue_payments"
)
plt.title("Overdue Payments vs Customer Churn")
plt.xlabel("Churn (0 = No, 1 = Yes)")
plt.ylabel("Overdue Payments")
plt.show()
```

Multivariate analysis

HEATMAP



```python
import matplotlib.pyplot as plt
import seaborn as sns
num_cols = [
    "monthly_minutes",
    "monthly_data_mb",
    "monthly_sms",
    "contract_months",
    "overdue_payments",
    "churn_next_month"
]
corr = df[num_cols].corr()
plt.figure(figsize=(10, 7))
sns.heatmap(
    corr,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)
plt.title("Multivariate Analysis - Correlation Heatmap")
plt.show()
```
